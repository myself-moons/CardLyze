```python
# Visa
import pandas as pd, ipywidgets as w
from IPython.display import display, clear_output

# ---------- CONFIG ----------
FILE = "/content/drive/MyDrive/Card_Data/Cards_with_Columns/visa_with_headers_28_10_2026.xlsx"
NUMERIC_COLS = ["SPTRN$", "sputa", "spfeam", "spfgst", "spmrkp", "spmgst", "spptam", "spmusa"]
DATE_COLS = {"spbusd": "%Y-%m-%d", "spdate": "%d%m%Y"}
BLANK, MAX_SHOWN = "<blank>", 1000
pd.set_option("display.max_rows", None, "display.max_columns", None)

# ---------- LOAD & TYPE ----------
df = pd.read_excel(FILE, dtype=str)
for c in NUMERIC_COLS:
    conv = pd.to_numeric(df[c], errors="coerce")
    bad = (conv.isna() & df[c].notna()).sum()
    if bad: print(f"WARNING {c}: {bad} non-numeric values turned into blanks")
    df[c] = conv
for c, fmt in DATE_COLS.items():
    conv = pd.to_datetime(df[c], format=fmt, errors="coerce")
    bad = (conv.isna() & df[c].notna()).sum()
    if bad: print(f"WARNING {c}: {bad} values did not match {fmt}, turned into blanks")
    df[c] = conv

# ---------- OVERVIEW ----------
print(f"\n{FILE}: {df.shape[0]:,} rows x {df.shape[1]} columns\n")
display(pd.DataFrame({"dtype": df.dtypes.astype(str), "non_null": df.notna().sum(),
                      "null_%": (df.isna().mean() * 100).round(1), "unique": df.nunique()}))
print("Numeric summary:");   display(df[NUMERIC_COLS].describe().T)
print("Date ranges:");       display(pd.DataFrame({c: [df[c].min(), df[c].max()] for c in DATE_COLS}, index=["min", "max"]).T)

# ---------- INTERACTIVE EXPLORER ----------
state = {"work": df.copy(), "filters": []}

def lab(s):
    if pd.api.types.is_datetime64_any_dtype(s):
        d = s.dropna()
        fmt = "%Y-%m-%d" if (d.dt.normalize() == d).all() else "%Y-%m-%d %H:%M:%S"
        out = s.dt.strftime(fmt)
    else:
        out = s.astype(object)
    return out.where(s.notna(), BLANK).astype(str)

def current():
    v = state["work"]; m = pd.Series(True, index=v.index)
    for col, mode, vals in state["filters"]:
        hit = lab(v[col]).isin(vals)
        m &= hit if mode == "keep" else ~hit
    return v[m]

col_dd  = w.Dropdown(options=[(f"{c}  ({df[c].nunique()} unique)", c) for c in df.columns], description="Column:", layout=w.Layout(width="520px"))
search  = w.Text(description="Search:", placeholder="type to narrow the value list", layout=w.Layout(width="520px"))
values  = w.SelectMultiple(options=[], rows=14, layout=w.Layout(width="720px"))
newval  = w.Text(description="Replace with:", placeholder="new value (empty = blank)", layout=w.Layout(width="520px"))
rows    = w.IntSlider(value=10, min=5, max=100, description="Preview rows:")
info, msg, out = w.Label(), w.Label(), w.Output()
btn = lambda t, s="": w.Button(description=t, button_style=s, layout=w.Layout(width="150px"))
b_all, b_keep, b_drop = btn("Select all shown"), btn("Keep selected", "success"), btn("Exclude selected", "warning")
b_undo, b_clear, b_reset = btn("Undo last filter"), btn("Clear filters"), btn("Reset data", "danger")
b_repl, b_exp = btn("Replace selected", "info"), btn("Export filtered")

def refresh_values(*_):
    vc = lab(current()[col_dd.value]).value_counts()
    q = search.value.strip().lower()
    if q: vc = vc[vc.index.str.lower().str.contains(q, regex=False)]
    info.value = f"{len(vc):,} unique values in current view" + (f" (showing top {MAX_SHOWN})" if len(vc) > MAX_SHOWN else "")
    values.options = [(f"{k}   [{n:,}]", k) for k, n in vc.head(MAX_SHOWN).items()]

def show():
    v = current()
    with out:
        clear_output(wait=True)
        print(f"Rows in view: {len(v):,} of {len(state['work']):,}")
        for i, (c, m, vals) in enumerate(state["filters"], 1):
            print(f"  {i}. {c} {'IN' if m == 'keep' else 'NOT IN'} {vals[:5]}{' ...' if len(vals) > 5 else ''}")
        display(v.head(rows.value))
    refresh_values()

def add_filter(mode):
    if values.value:
        state["filters"].append((col_dd.value, mode, list(values.value))); msg.value = ""; show()

def on_replace(_):
    if not values.value: return
    col, s, new = col_dd.value, state["work"][col_dd.value], newval.value.strip()
    try:
        if new == "": nv = pd.NA if not pd.api.types.is_numeric_dtype(s) else float("nan")
        elif pd.api.types.is_numeric_dtype(s): nv = float(new)
        elif pd.api.types.is_datetime64_any_dtype(s): nv = pd.to_datetime(new)
        else: nv = new
    except Exception as e:
        msg.value = f"Cannot convert '{new}' for this column type: {e}"; return
    hit = lab(s).isin(list(values.value))
    state["work"].loc[hit, col] = nv
    state["filters"] = [f for f in state["filters"] if f[0] != col]
    msg.value = f"Replaced {hit.sum():,} rows in {col}"; show()

def on_export(_):
    name = f"filtered_{FILE}"; current().to_excel(name, index=False); msg.value = f"Saved {name}"
    try:
        from google.colab import files; files.download(name)
    except Exception: pass

def reset(_):
    state["work"] = df.copy(); state["filters"].clear(); msg.value = ""; show()

col_dd.observe(refresh_values, "value"); search.observe(refresh_values, "value"); rows.observe(lambda _: show(), "value")
b_all.on_click(lambda _: setattr(values, "value", tuple(v for _, v in values.options)))
b_keep.on_click(lambda _: add_filter("keep")); b_drop.on_click(lambda _: add_filter("drop"))
b_undo.on_click(lambda _: (state["filters"].pop() if state["filters"] else None, show()))
b_clear.on_click(lambda _: (state["filters"].clear(), show())); b_reset.on_click(reset)
b_repl.on_click(on_replace); b_exp.on_click(on_export)

display(w.VBox([col_dd, search, info, values, w.HBox([b_all, b_keep, b_drop]), newval,
                w.HBox([b_repl, b_undo, b_clear, b_reset, b_exp]), msg, rows, out]))
show()
```
